```python
import numpy as np
import matplotlib.pyplot as plt

# Extraigo los datos de las velocidades de los ficheros
Velocidades_1 = np.loadtxt("Velocidades_Atomos_v_1_ej_3.txt")
Velocidades_1_5 = np.loadtxt("Velocidades_Atomos_v_1_5_ej_3.txt")
Velocidades_2 = np.loadtxt("Velocidades_Atomos_v_2_ej_3.txt")
Velocidades_2_5 = np.loadtxt("Velocidades_Atomos_v_2_5_ej_3.txt")
Velocidades_3 = np.loadtxt("Velocidades_Atomos_v_3_ej_3.txt")

# Extraigo las velocidades v_x e v_y de los átomos entre t = 100 y t = 250 segundos, donde ya se ha alcanzado el equilibrio 
v_x_1 = Velocidades_1[50000:, 0::2]
v_y_1 = Velocidades_1[50000:, 1::2]
v_x_1_5 = Velocidades_1_5[50000:, 0::2]
v_y_1_5 = Velocidades_1_5[50000:, 1::2]
v_x_2 = Velocidades_2[50000:, 0::2]
v_y_2 = Velocidades_2[50000:, 1::2]
v_x_2_5 = Velocidades_2_5[50000:, 0::2]
v_y_2_5 = Velocidades_2_5[50000:, 1::2]
v_x_3 = Velocidades_3[50000:, 0::2]
v_y_3 = Velocidades_3[50000:, 1::2]

def T_media(v_x, v_y):

    # Hago la media sobre las columnas, es decir, sobre los átomos, para obtener la temperatura en cada instante de tiempo 
    T = 0.5 * np.mean(v_x**2 + v_y**2, axis = 1) 

    # Hago la media sobre el tiempo para obtener la temperatura media
    T = np.mean(T)

    return T

# Calculo la temperatura media en función del tiempo para cada velocidad inicial
T_1 = T_media(v_x_1, v_y_1)
T_1_5 = T_media(v_x_1_5, v_y_1_5)
T_2 = T_media(v_x_2, v_y_2)
T_2_5 = T_media(v_x_2_5, v_y_2_5)
T_3 = T_media(v_x_3, v_y_3)

# Almaceno las temperaturas medias en un array
T = np.array([T_1, T_1_5, T_2, T_2_5, T_3])

# Extraigo los datos de la presión de las partículas en las paredes de la caja
Presion_1 = np.loadtxt("Presion_v_1_ej_3.txt")
Presion_1_5 = np.loadtxt("Presion_v_1_5_ej_3.txt")
Presion_2 = np.loadtxt("Presion_v_2_ej_3.txt")
Presion_2_5 = np.loadtxt("Presion_v_2_5_ej_3.txt")
Presion_3 = np.loadtxt("Presion_v_3_ej_3.txt")

# Calculo la presión media cuando se ha alncazado el equilibrio, es decir, entre t = 100 y t = 250 segundos
Presion_1_media = np.mean(Presion_1[50000:])
Presion_1_5_media = np.mean(Presion_1_5[50000:])
Presion_2_media = np.mean(Presion_2[50000:])
Presion_2_5_media = np.mean(Presion_2_5[50000:])
Presion_3_media = np.mean(Presion_3[50000:])

# Almaceno las presiones medias en un array
Presion = np.array([Presion_1_media, Presion_1_5_media, Presion_2_media, Presion_2_5_media, Presion_3_media])

# Hago el gráfico de la presión en función de la temperatura
fig, ax = plt.subplots(figsize = (5, 5))

ax.plot(T, Presion, color = "red", marker = "o")
ax.set_xlabel("T")
ax.set_ylabel("P")

fig.savefig("Presion_vs_Temperatura_ej_3.png", dpi = 300)
plt.show()
```
