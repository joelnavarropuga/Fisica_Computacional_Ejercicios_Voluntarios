Calculo las fluctuaciones de la distancia entre dos partículas.

\begin{equation}
\nonumber

\langle (r_{ij}(t))^2\rangle = \langle (r_i(t) - r_j(t))^2 \rangle

\end{equation}

```python
import numpy as np
import matplotlib.pyplot as plt

# Cargo los datos de las posiciones del átomo 0 desde el fichero generado, y las velocidades de los demás
Posicion_atomo = np.loadtxt("Posicion_Par_Atomos_ej_7.txt")
Velocidades = np.loadtxt("Velocidades_Atomos_ej_7.txt")

# Defino el paso temporal y la longitud de la caja
h = 0.002
L = 12

# Extraigo las posiciones x e y de los átomos 0 y 1, sin aplicar las condiciones periódicas
x_0 = Posicion_atomo[0:, 0]
y_0 = Posicion_atomo[0:, 1]

x_1 = Posicion_atomo[0:, 2]
y_1 = Posicion_atomo[0:, 3]

# Calculo la distancia entre ambos átomos
dx = x_1 - x_0
dy = y_1 - y_0

r_cuadrado = dx**2 + dy**2

# Realizo los promedios en los intervalos de tiempo correspondientes a los reescalados de las velocidades, es decir, a t = 500, 1000, 2000 y 3000 segundos
r_cuadrado_0 = np.mean(r_cuadrado[:250000])
r_cuadrado_500 = np.mean(r_cuadrado[250000:500000])
r_cuadrado_1000 = np.mean(r_cuadrado[500000:1000000])
r_cuadrado_2000 = np.mean(r_cuadrado[1000000:1500000])
r_cuadrado_3000 = np.mean(r_cuadrado[1500000:2000000])

# Almaceno en un array
r_cuadrado_medio = np.array([r_cuadrado_0, r_cuadrado_500, r_cuadrado_1000, r_cuadrado_2000, r_cuadrado_3000])

# Extraigo las velocidades v_x e v_y de los átomos
v_x = Velocidades[0:, 0::2]
v_y = Velocidades[0:, 1::2]

# Hago la media sobre las columnas, es decir, sobre los átomos, para obtener la temperatura en cada instante de tiempo 
T = 0.5 * np.mean(v_x**2 + v_y**2, axis = 1)

# Hago la media sobre los intervalos de tiempo correspondientes a los reescalados de las velocidades, es decir, a t = 500, 1000, 2000 y 3000 segundos
T_0 = np.mean(T[:250000])
T_500 = np.mean(T[250000:500000])
T_1000 = np.mean(T[500000:1000000])
T_2000 = np.mean(T[1000000:1500000])
T_3000 = np.mean(T[1500000:2000000])

# Almaceno en un array
T_medio = np.array([T_0, T_500, T_1000, T_2000, T_3000])

# Grafico la evolución de r_cuadrado en función de la temperatura T
fig, ax = plt.subplots(figsize=(5, 5))

ax.plot(T_medio, r_cuadrado_medio, marker = "o", color = 'red')
ax.set_xlabel("T")
ax.set_ylabel(r"$\langle(r_i(t) - r_j(t))^2 \rangle$")
ax.set_title("Separación cuadrática media entre un par\n de átomos en función de la temperatura")

fig.savefig("r_vs_t_ej_7.png", dpi = 150)
plt.show()

fig.savefig("r_vs_t_ej_7.png", dpi = 150)
plt.show()

```
