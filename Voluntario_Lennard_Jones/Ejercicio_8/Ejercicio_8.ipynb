```python
import numpy as np
import matplotlib.pyplot as plt

# Cargo los histogramas
hist_solido = np.loadtxt('Histograma_solido_ej_8.txt')
hist_liquido = np.loadtxt('Histograma_liquido_ej_8.txt')
hist_gas = np.loadtxt('Histograma_gas_ej_8.txt')

# Defino los radios correspondientes a cada bin del histograma
r = np.linspace(1.0, 3.0, 40)

# Normalizo el histograma para obtener la función de distribución de probabilidad
g_solido = hist_solido / (r * np.sum(hist_solido))
g_liquido = hist_liquido / (r * np.sum(hist_liquido))
g_gas = hist_gas / (r * np.sum(hist_gas))

# Grafico el histograma
fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(r, g_solido, color='blue', label = 'Estado sólido', marker = 'o', markersize = 3)
ax.plot(r, g_liquido, color='red', label = 'Estado líquido', marker = 'o', markersize = 3)
ax.plot(r, g_gas, color='green', label = 'Estado gaseoso', marker = 'o', markersize = 3)
ax.set_xlabel('r')
ax.set_ylabel('g(r)')
ax.legend()

fig.savefig('Funcion_de_correlacion_ej_8.png', dpi=300)
```
