Represento el solapamiento para distintas temperaturas, de manera que el valor de solapamiento asociado a una temperatura es el solapamiento tras 20 pasos Monte Carlo, para que de esa forma le haya dado tiempo al sistema a evolucionar y alcanzar la situación de equilibrio.

```python
import numpy as np
import matplotlib.pyplot as plt

# Cargo las configuraciones de las neuronas desde el archivo generado
Matriz_neuronas_inicial_aleatoria_T_0_0001 = np.loadtxt('configuracion_inicial_neuronas_aleatoria.txt')
Matriz_neuronas_inicial_deformada_T_0_0001 = np.loadtxt('configuracion_inicial_neuronas_deformada.txt')

Matriz_neuronas_inicial_aleatoria_T_0_001 = np.loadtxt('configuracion_inicial_neuronas_aleatoria_T_0.001000.txt')
Matriz_neuronas_inicial_deformada_T_0_001 = np.loadtxt('configuracion_inicial_neuronas_deformada_T_0.001000.txt')

Matriz_neuronas_inicial_aleatoria_T_0_01 = np.loadtxt('configuracion_inicial_neuronas_aleatoria_T_0.010000.txt')
Matriz_neuronas_inicial_deformada_T_0_01 = np.loadtxt('configuracion_inicial_neuronas_deformada_T_0.010000.txt')

Matriz_neuronas_inicial_aleatoria_T_0_05 = np.loadtxt('configuracion_inicial_neuronas_aleatoria_T_0.050000.txt')
Matriz_neuronas_inicial_deformada_T_0_05 = np.loadtxt('configuracion_inicial_neuronas_deformada_T_0.050000.txt')

Matriz_neuronas_inicial_aleatoria_T_0_06 = np.loadtxt('configuracion_inicial_neuronas_aleatoria_T_0.060000.txt')
Matriz_neuronas_inicial_deformada_T_0_06 = np.loadtxt('configuracion_inicial_neuronas_deformada_T_0.060000.txt')

Matriz_neuronas_inicial_aleatoria_T_0_07 = np.loadtxt('configuracion_inicial_neuronas_aleatoria_T_0.070000.txt')
Matriz_neuronas_inicial_deformada_T_0_07 = np.loadtxt('configuracion_inicial_neuronas_deformada_T_0.070000.txt')

Matriz_neuronas_inicial_aleatoria_T_0_08 = np.loadtxt('configuracion_inicial_neuronas_aleatoria_T_0.080000.txt')
Matriz_neuronas_inicial_deformada_T_0_08 = np.loadtxt('configuracion_inicial_neuronas_deformada_T_0.080000.txt')

Matriz_neuronas_inicial_aleatoria_T_0_09 = np.loadtxt('configuracion_inicial_neuronas_aleatoria_T_0.090000.txt')
Matriz_neuronas_inicial_deformada_T_0_09 = np.loadtxt('configuracion_inicial_neuronas_deformada_T_0.090000.txt')

Matriz_neuronas_inicial_aleatoria_T_0_1 = np.loadtxt('configuracion_inicial_neuronas_aleatoria_T_0.100000.txt')
Matriz_neuronas_inicial_deformada_T_0_1 = np.loadtxt('configuracion_inicial_neuronas_deformada_T_0.100000.txt')

# Modifico la forma de la matriz para que sea un array de M matrices 150x150
Matriz_neuronas_inicial_aleatoria_T_0_0001 = Matriz_neuronas_inicial_aleatoria_T_0_0001.reshape(-1, 150, 150)
Matriz_neuronas_inicial_deformada_T_0_0001 = Matriz_neuronas_inicial_deformada_T_0_0001.reshape(-1, 150, 150)
Matriz_neuronas_inicial_aleatoria_T_0_001 = Matriz_neuronas_inicial_aleatoria_T_0_001.reshape(-1, 150, 150)
Matriz_neuronas_inicial_deformada_T_0_001 = Matriz_neuronas_inicial_deformada_T_0_001.reshape(-1, 150, 150)
Matriz_neuronas_inicial_aleatoria_T_0_01 = Matriz_neuronas_inicial_aleatoria_T_0_01.reshape(-1, 150, 150)
Matriz_neuronas_inicial_deformada_T_0_01 = Matriz_neuronas_inicial_deformada_T_0_01.reshape(-1, 150, 150)
Matriz_neuronas_inicial_aleatoria_T_0_05 = Matriz_neuronas_inicial_aleatoria_T_0_05.reshape(-1, 150, 150)
Matriz_neuronas_inicial_deformada_T_0_05 = Matriz_neuronas_inicial_deformada_T_0_05.reshape(-1, 150, 150)
Matriz_neuronas_inicial_aleatoria_T_0_06 = Matriz_neuronas_inicial_aleatoria_T_0_06.reshape(-1, 150, 150)
Matriz_neuronas_inicial_deformada_T_0_06 = Matriz_neuronas_inicial_deformada_T_0_06.reshape(-1, 150, 150)
Matriz_neuronas_inicial_aleatoria_T_0_07 = Matriz_neuronas_inicial_aleatoria_T_0_07.reshape(-1, 150, 150)
Matriz_neuronas_inicial_deformada_T_0_07 = Matriz_neuronas_inicial_deformada_T_0_07.reshape(-1, 150, 150)
Matriz_neuronas_inicial_aleatoria_T_0_08 = Matriz_neuronas_inicial_aleatoria_T_0_08.reshape(-1, 150, 150)
Matriz_neuronas_inicial_deformada_T_0_08 = Matriz_neuronas_inicial_deformada_T_0_08.reshape(-1, 150, 150)
Matriz_neuronas_inicial_aleatoria_T_0_09 = Matriz_neuronas_inicial_aleatoria_T_0_09.reshape(-1, 150, 150)
Matriz_neuronas_inicial_deformada_T_0_09 = Matriz_neuronas_inicial_deformada_T_0_09.reshape(-1, 150, 150)
Matriz_neuronas_inicial_aleatoria_T_0_1 = Matriz_neuronas_inicial_aleatoria_T_0_1.reshape(-1, 150, 150)
Matriz_neuronas_inicial_deformada_T_0_1 = Matriz_neuronas_inicial_deformada_T_0_1.reshape(-1, 150, 150)

M = Matriz_neuronas_inicial_aleatoria_T_0_0001.shape[0]
N = Matriz_neuronas_inicial_aleatoria_T_0_0001.shape[1]

# Cargo el patrón original
patron_original = np.loadtxt('Patron_original.txt')

# Calculo el valor de a
a = np.mean(patron_original)

# Defino la función que calcula el solapamiento
def solapamiento(patron_original, s, a, N):

    m = 0.0
    for i in range(N):
        for j in range(N):
                
            m += (patron_original[i, j] - a) * (s[i, j] - a) ;
        
    m = m / ((N * N) * a * (1 - a)) 
    return m

# Calculo el solapamiento para cada configuración de las neuronas al cabo de 20 pasos Monte Carlo para cada temperatura

solapamiento_configuracion_inicial_aleatoria_T_0_0001 = solapamiento(patron_original, Matriz_neuronas_inicial_aleatoria_T_0_0001[-1], a, N)
solapamiento_configuracion_inicial_deformada_T_0_0001 = solapamiento(patron_original, Matriz_neuronas_inicial_deformada_T_0_0001[-1], a, N)
solapamiento_configuracion_inicial_aleatoria_T_0_001 = solapamiento(patron_original, Matriz_neuronas_inicial_aleatoria_T_0_001[-1], a, N)
solapamiento_configuracion_inicial_deformada_T_0_001 = solapamiento(patron_original, Matriz_neuronas_inicial_deformada_T_0_001[-1], a, N)
solapamiento_configuracion_inicial_aleatoria_T_0_01 = solapamiento(patron_original, Matriz_neuronas_inicial_aleatoria_T_0_01[-1], a, N)
solapamiento_configuracion_inicial_deformada_T_0_01 = solapamiento(patron_original, Matriz_neuronas_inicial_deformada_T_0_01[-1], a, N)
solapamiento_configuracion_inicial_aleatoria_T_0_05 = solapamiento(patron_original, Matriz_neuronas_inicial_aleatoria_T_0_05[-1], a, N)
solapamiento_configuracion_inicial_deformada_T_0_05 = solapamiento(patron_original, Matriz_neuronas_inicial_deformada_T_0_05[-1], a, N)
solapamiento_configuracion_inicial_aleatoria_T_0_06 = solapamiento(patron_original, Matriz_neuronas_inicial_aleatoria_T_0_06[-1], a, N)
solapamiento_configuracion_inicial_deformada_T_0_06 = solapamiento(patron_original, Matriz_neuronas_inicial_deformada_T_0_06[-1], a, N)
solapamiento_configuracion_inicial_aleatoria_T_0_07 = solapamiento(patron_original, Matriz_neuronas_inicial_aleatoria_T_0_07[-1], a, N)
solapamiento_configuracion_inicial_deformada_T_0_07 = solapamiento(patron_original, Matriz_neuronas_inicial_deformada_T_0_07[-1], a, N)
solapamiento_configuracion_inicial_aleatoria_T_0_08 = solapamiento(patron_original, Matriz_neuronas_inicial_aleatoria_T_0_08[-1], a, N)
solapamiento_configuracion_inicial_deformada_T_0_08 = solapamiento(patron_original, Matriz_neuronas_inicial_deformada_T_0_08[-1], a, N)
solapamiento_configuracion_inicial_aleatoria_T_0_09 = solapamiento(patron_original, Matriz_neuronas_inicial_aleatoria_T_0_09[-1], a, N)
solapamiento_configuracion_inicial_deformada_T_0_09 = solapamiento(patron_original, Matriz_neuronas_inicial_deformada_T_0_09[-1], a, N)
solapamiento_configuracion_inicial_aleatoria_T_0_1 = solapamiento(patron_original, Matriz_neuronas_inicial_aleatoria_T_0_1[-1], a, N)
solapamiento_configuracion_inicial_deformada_T_0_1 = solapamiento(patron_original, Matriz_neuronas_inicial_deformada_T_0_1[-1], a, N)

# Guardo los solapamientos en un array
m_aleatorio = np.array([solapamiento_configuracion_inicial_aleatoria_T_0_0001, solapamiento_configuracion_inicial_aleatoria_T_0_001, solapamiento_configuracion_inicial_aleatoria_T_0_01, solapamiento_configuracion_inicial_aleatoria_T_0_05, solapamiento_configuracion_inicial_aleatoria_T_0_06, solapamiento_configuracion_inicial_aleatoria_T_0_07, solapamiento_configuracion_inicial_aleatoria_T_0_08, solapamiento_configuracion_inicial_aleatoria_T_0_09, solapamiento_configuracion_inicial_aleatoria_T_0_1])
m_deformada = np.array([solapamiento_configuracion_inicial_deformada_T_0_0001, solapamiento_configuracion_inicial_deformada_T_0_001, solapamiento_configuracion_inicial_deformada_T_0_01, solapamiento_configuracion_inicial_deformada_T_0_05, solapamiento_configuracion_inicial_deformada_T_0_06, solapamiento_configuracion_inicial_deformada_T_0_07, solapamiento_configuracion_inicial_deformada_T_0_08, solapamiento_configuracion_inicial_deformada_T_0_09, solapamiento_configuracion_inicial_deformada_T_0_1])

# Creo un vector que contiene las temperaturas para graficar el solapamiento en función de la temperatura al cabo de 20 pasos Monte Carlo
T = np.array([0.0001, 0.001, 0.01, 0.05, 0.06, 0.07, 0.08, 0.09, 0.1])

# Defino la figura donde voy a graficar
fig, ax = plt.subplots(1,2, figsize = (12, 5))

ax[0].plot(T, m_aleatorio, marker = 'o', color = "red")
ax[1].plot(T, m_deformada, marker = 'o', color = "blue")

ax[0].set_title("Solapamiento en función de la temperatura\n para una configuración inicial aleatoria")
ax[0].set_xlabel("T")
ax[0].set_ylabel("m")

ax[1].set_title("Solapamiento en función de la temperatura\n para una configuración inicial deformada")
ax[1].set_xlabel("T")
ax[1].set_ylabel("m")

fig.savefig("Solapamiento_2.png", dpi = 300)
plt.show()
```
