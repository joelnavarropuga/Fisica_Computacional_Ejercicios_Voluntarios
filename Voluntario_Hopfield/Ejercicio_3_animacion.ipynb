Animación de la evolución de la red neuronal para los patrones iniciales aleatorio y deformado, a una temperatura $T$ = $10^{-4}$, cuando la red posee varios patrones almacenados, concretamente los dígitos 0, 1 y 2.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation

# Cargo las configuraciones de las neuronas desde el archivo generado
Matriz_neuronas_inicial_aleatoria = np.loadtxt('configuracion_inicial_neuronas_aleatoria_mu.txt')
Matriz_neuronas_inicial_deformada = np.loadtxt('configuracion_inicial_neuronas_deformada_mu.txt')

# Modifico la forma de la matriz para que sea un array de M matrices 150x150
N = 30
Matriz_neuronas_inicial_aleatoria = Matriz_neuronas_inicial_aleatoria.reshape(-1, N, N)
Matriz_neuronas_inicial_deformada = Matriz_neuronas_inicial_deformada.reshape(-1, N, N)

M = Matriz_neuronas_inicial_aleatoria.shape[0]
    
# Creo la figura para la animación
fig, ax = plt.subplots(1,2,figsize = (12, 5))
ax[0].set_title('Evolución de las neuronas\n con patrón inicial aleatorio')
ax[0].set_xlim(0, N - 1)
ax[0].set_ylim(0, N - 1)
ax[1].set_title('Evolución de las neuronas\n con patrón inicial 0 deformado')
ax[1].set_xlim(0, N - 1)
ax[1].set_ylim(0, N - 1)

# Configuro la línea para la animación, donde los 1 se representan en color negro y los 0 se dejan en blanco
line_1, = ax[0].plot([], [], color = 'black', marker = 's', linestyle = 'None', markersize = 9.5)
line_2, = ax[1].plot([], [], color = 'black', marker = 's', linestyle = 'None', markersize = 9.5)

# Función para realizar la animación
def animate(i):
    
    pos_1 = np.where(Matriz_neuronas_inicial_aleatoria[i, :, :] == 1)
    pos_2 = np.where(Matriz_neuronas_inicial_deformada[i, :, :] == 1)

    line_1.set_data(pos_1[1], N - 1 - pos_1[0])
    line_2.set_data(pos_2[1], N - 1 - pos_2[0])

    return line_1, line_2

# Creo una animación usando la función animate
ani = animation.FuncAnimation(fig, animate, frames = M, interval = 100, blit = True)

# Guardo la animación como un archivo GIF
ani.save("Configuracion_neuronas_mu_animacion.gif", writer='pillow', dpi = 150)
```
