Solapamiento en función del paso Monte Carlo para la temperatura T = $10^{-4}$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Cargo las configuraciones de las neuronas desde el archivo generado
Matriz_neuronas_inicial_aleatoria = np.loadtxt('Configuracion_inicial_neuronas_aleatoria_mu.txt')
Matriz_neuronas_inicial_deformada = np.loadtxt('Configuracion_inicial_neuronas_deformada_mu.txt')

N = Matriz_neuronas_inicial_aleatoria.shape[1]

# Modifico la forma de la matriz para que sea un array de M matrices 30x30
Matriz_neuronas_inicial_aleatoria = Matriz_neuronas_inicial_aleatoria.reshape(-1, N, N)
Matriz_neuronas_inicial_deformada = Matriz_neuronas_inicial_deformada.reshape(-1, N, N)

M = Matriz_neuronas_inicial_aleatoria.shape[0]

# Cargo los patrones originales
patron_original_0 = np.loadtxt('0.txt')
patron_original_1 = np.loadtxt('1.txt')
patron_original_2 = np.loadtxt('2.txt')

# Calculo el valor de a
def a(patron_original):
    return np.mean(patron_original)

# Defino la función que calcula el solapamiento
def solapamiento(patron_original, s, a, N):

    m = 0.0
    for i in range(N):
        for j in range(N):
                
            m += (patron_original[i, j] - a) * (s[i, j] - a) ;
        
    m = m / ((N * N) * a * (1 - a)) # Pongo el valor absoluto para que no influya que la imagen se ha invertido
    return m

# Calculo el solapamiento para cada configuración de las neuronas
solapamiento_0_aleatoria = np.zeros(M)
solapamiento_0_deformada = np.zeros(M)

solapamiento_1_aleatoria = np.zeros(M)
solapamiento_1_deformada = np.zeros(M)

solapamiento_2_aleatoria = np.zeros(M)
solapamiento_2_deformada = np.zeros(M)

for k in range(M):
    solapamiento_0_aleatoria[k] = solapamiento(patron_original_0, Matriz_neuronas_inicial_aleatoria[k], a(patron_original_0), N)
    solapamiento_0_deformada[k] = solapamiento(patron_original_0, Matriz_neuronas_inicial_deformada[k], a(patron_original_0), N)
    solapamiento_1_aleatoria[k] = solapamiento(patron_original_1, Matriz_neuronas_inicial_aleatoria[k], a(patron_original_1), N)
    solapamiento_1_deformada[k] = solapamiento(patron_original_1, Matriz_neuronas_inicial_deformada[k], a(patron_original_1), N)
    solapamiento_2_aleatoria[k] = solapamiento(patron_original_2, Matriz_neuronas_inicial_aleatoria[k], a(patron_original_2), N)
    solapamiento_2_deformada[k] = solapamiento(patron_original_2, Matriz_neuronas_inicial_deformada[k], a(patron_original_2), N)

# Creo un vector que contiene los números del 0 al M-1 para graficar el solapamiento en función del número del paso Monte Carlo
paso_Monte_Carlo = np.arange(M)

# Defino la figura donde voy a graficar
fig, ax = plt.subplots(1, 2, figsize = (12, 4.5))
ax[0].plot(paso_Monte_Carlo, solapamiento_0_aleatoria, marker = 'o', color = "red", label = "Solapamiento con 0")
ax[0].plot(paso_Monte_Carlo, solapamiento_1_aleatoria, marker = 'o', color = "blue", label = "Solapamiento con 1")
ax[0].plot(paso_Monte_Carlo, solapamiento_2_aleatoria, marker = 'o', color = "green", label = "Solapamiento con 2")

ax[1].plot(paso_Monte_Carlo, solapamiento_0_deformada, marker = 'o', color = "red", label = "Solapamiento con 0")
ax[1].plot(paso_Monte_Carlo, solapamiento_1_deformada, marker = 'o', color = "blue", label = "Solapamiento con 1")
ax[1].plot(paso_Monte_Carlo, solapamiento_2_deformada, marker = 'o', color = "green", label = "Solapamiento con 2")

ax[0].set_title("Solapamiento en función del tiempo\n para un patrón inicial aleatorio")
ax[0].set_xlabel("Paso Monte Carlo")
ax[0].set_ylabel("m")
ax[0].legend()

ax[1].set_title("Solapamiento en función del tiempo\n para el patrón inicial 0 deformado")
ax[1].set_xlabel("Paso Monte Carlo")
ax[1].set_ylabel("m")
ax[1].legend(loc = "lower right")

fig.savefig("Solapamiento_3.png", dpi = 300)
plt.show()
```
