```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation

# Cargo las configuraciones de las neuronas para unas determiandas temperaturas
Matriz_neuronas_inicial_aleatoria_T_0_0001 = np.loadtxt('Configuracion_inicial_neuronas_aleatoria.txt')
Matriz_neuronas_inicial_deformada_T_0_0001 = np.loadtxt('Configuracion_inicial_neuronas_deformada.txt')

Matriz_neuronas_inicial_aleatoria_T_0_01 = np.loadtxt('Configuracion_inicial_neuronas_aleatoria_T_0.010000.txt')
Matriz_neuronas_inicial_deformada_T_0_01 = np.loadtxt('Configuracion_inicial_neuronas_deformada_T_0.010000.txt')

Matriz_neuronas_inicial_aleatoria_T_0_05 = np.loadtxt('Configuracion_inicial_neuronas_aleatoria_T_0.050000.txt')
Matriz_neuronas_inicial_deformada_T_0_05 = np.loadtxt('Configuracion_inicial_neuronas_deformada_T_0.050000.txt')

Matriz_neuronas_inicial_aleatoria_T_0_08 = np.loadtxt('Configuracion_inicial_neuronas_aleatoria_T_0.080000.txt')
Matriz_neuronas_inicial_deformada_T_0_08 = np.loadtxt('Configuracion_inicial_neuronas_deformada_T_0.080000.txt')

Matriz_neuronas_inicial_aleatoria_T_0_1 = np.loadtxt('Configuracion_inicial_neuronas_aleatoria_T_0.100000.txt')
Matriz_neuronas_inicial_deformada_T_0_1 = np.loadtxt('Configuracion_inicial_neuronas_deformada_T_0.100000.txt')

# Modifico la forma de la matriz para que sea un array de M matrices 150x150
Matriz_neuronas_inicial_aleatoria_T_0_0001 = Matriz_neuronas_inicial_aleatoria_T_0_0001.reshape(-1, 150, 150)
Matriz_neuronas_inicial_deformada_T_0_0001 = Matriz_neuronas_inicial_deformada_T_0_0001.reshape(-1, 150, 150)
Matriz_neuronas_inicial_aleatoria_T_0_01 = Matriz_neuronas_inicial_aleatoria_T_0_01.reshape(-1, 150, 150)
Matriz_neuronas_inicial_deformada_T_0_01 = Matriz_neuronas_inicial_deformada_T_0_01.reshape(-1, 150, 150)
Matriz_neuronas_inicial_aleatoria_T_0_05 = Matriz_neuronas_inicial_aleatoria_T_0_05.reshape(-1, 150, 150)
Matriz_neuronas_inicial_deformada_T_0_05 = Matriz_neuronas_inicial_deformada_T_0_05.reshape(-1, 150, 150)
Matriz_neuronas_inicial_aleatoria_T_0_08 = Matriz_neuronas_inicial_aleatoria_T_0_08.reshape(-1, 150, 150)
Matriz_neuronas_inicial_deformada_T_0_08 = Matriz_neuronas_inicial_deformada_T_0_08.reshape(-1, 150, 150)
Matriz_neuronas_inicial_aleatoria_T_0_1 = Matriz_neuronas_inicial_aleatoria_T_0_1.reshape(-1, 150, 150)
Matriz_neuronas_inicial_deformada_T_0_1 = Matriz_neuronas_inicial_deformada_T_0_1.reshape(-1, 150, 150)

M = Matriz_neuronas_inicial_aleatoria_T_0_0001.shape[0]

# Creo la figura para la animación
fig, ax = plt.subplots(5,2,figsize = (10, 25))

fig.subplots_adjust(hspace=0.35, wspace=0.25)

ax[0,0].set_title('Evolución de las neuronas con patrón\n inicial aleatorio a T = 0.0001')
ax[0,1].set_title('Evolución de las neuronas con patrón\n inicial deformado a T = 0.0001')
ax[1,0].set_title('Evolución de las neuronas con patrón\n inicial aleatorio a T = 0.01')
ax[1,1].set_title('Evolución de las neuronas con patrón\n inicial deformado a T = 0.01')
ax[2,0].set_title('Evolución de las neuronas con patrón\n inicial aleatorio a T = 0.05')
ax[2,1].set_title('Evolución de las neuronas con patrón\n inicial deformado a T = 0.05')
ax[3,0].set_title('Evolución de las neuronas con patrón\n inicial aleatorio a T = 0.08')
ax[3,1].set_title('Evolución de las neuronas con patrón\n inicial deformado a T = 0.08')
ax[4,0].set_title('Evolución de las neuronas con patrón\n inicial aleatorio a T = 0.1')
ax[4,1].set_title('Evolución de las neuronas con patrón\n inicial deformado a T = 0.1')

for i in range(0, 5):
    for j in range(0, 2):
        ax[i,j].set_xlim(0, 149)
        ax[i,j].set_ylim(0, 149)

# Configuro la línea para la animación, donde los 1 se representan en color negro y los 0 se dejan en blanco
line_1, = ax[0,0].plot([], [], color = 'black', marker = 's', linestyle = 'None', markersize = 1.5)
line_2, = ax[0,1].plot([], [], color = 'black', marker = 's', linestyle = 'None', markersize = 1.5)
line_3, = ax[1,0].plot([], [], color = 'black', marker = 's', linestyle = 'None', markersize = 1.5)
line_4, = ax[1,1].plot([], [], color = 'black', marker = 's', linestyle = 'None', markersize = 1.5)
line_5, = ax[2,0].plot([], [], color = 'black', marker = 's', linestyle = 'None', markersize = 1.5)
line_6, = ax[2,1].plot([], [], color = 'black', marker = 's', linestyle = 'None', markersize = 1.5)
line_7, = ax[3,0].plot([], [], color = 'black', marker = 's', linestyle = 'None', markersize = 1.5)
line_8, = ax[3,1].plot([], [], color = 'black', marker = 's', linestyle = 'None', markersize = 1.5)
line_9, = ax[4,0].plot([], [], color = 'black', marker = 's', linestyle = 'None', markersize = 1.5)
line_10, = ax[4,1].plot([], [], color = 'black', marker = 's', linestyle = 'None', markersize = 1.5)


def animate(i):

    pos_1 = np.where(Matriz_neuronas_inicial_aleatoria_T_0_0001[i, :, :] == 1)
    pos_2 = np.where(Matriz_neuronas_inicial_deformada_T_0_0001[i, :, :] == 1)
    pos_3 = np.where(Matriz_neuronas_inicial_aleatoria_T_0_01[i, :, :] == 1)
    pos_4 = np.where(Matriz_neuronas_inicial_deformada_T_0_01[i, :, :] == 1)
    pos_5 = np.where(Matriz_neuronas_inicial_aleatoria_T_0_05[i, :, :] == 1)
    pos_6 = np.where(Matriz_neuronas_inicial_deformada_T_0_05[i, :, :] == 1)
    pos_7 = np.where(Matriz_neuronas_inicial_aleatoria_T_0_08[i, :, :] == 1)
    pos_8 = np.where(Matriz_neuronas_inicial_deformada_T_0_08[i, :, :] == 1)
    pos_9 = np.where(Matriz_neuronas_inicial_aleatoria_T_0_1[i, :, :] == 1)
    pos_10 = np.where(Matriz_neuronas_inicial_deformada_T_0_1[i, :, :] == 1)

    line_1.set_data(pos_1[1], 149 - pos_1[0])
    line_2.set_data(pos_2[1], 149 - pos_2[0])
    line_3.set_data(pos_3[1], 149 - pos_3[0])
    line_4.set_data(pos_4[1], 149 - pos_4[0])
    line_5.set_data(pos_5[1], 149 - pos_5[0])
    line_6.set_data(pos_6[1], 149 - pos_6[0])
    line_7.set_data(pos_7[1], 149 - pos_7[0])
    line_8.set_data(pos_8[1], 149 - pos_8[0])
    line_9.set_data(pos_9[1], 149 - pos_9[0])
    line_10.set_data(pos_10[1], 149 - pos_10[0])

    return line_1, line_2, line_3, line_4, line_5, line_6, line_7, line_8, line_9, line_10

# Creo una animación usando la función animate
ani = animation.FuncAnimation(fig, animate, frames = M, interval = 100, blit = True)

# Guardo la animación como un archivo GIF
ani.save("Configuracion_neuronas_animacion_T.gif", writer='pillow', dpi = 150)
```
