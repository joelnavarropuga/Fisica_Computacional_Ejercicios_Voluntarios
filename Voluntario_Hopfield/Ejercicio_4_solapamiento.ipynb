```python
import numpy as np
import matplotlib.pyplot as plt

# Cargo las configuraciones de las neuronas desde el archivo generado
Matriz_neuronas_0_deformada = np.loadtxt('Ej_4_10_patrones_numero_0.txt')
Matriz_neuronas_1_deformada = np.loadtxt('Ej_4_10_patrones_numero_1.txt')
Matriz_neuronas_2_deformada = np.loadtxt('Ej_4_10_patrones_numero_2.txt')
Matriz_neuronas_3_deformada = np.loadtxt('Ej_4_10_patrones_numero_3.txt')
Matriz_neuronas_4_deformada = np.loadtxt('Ej_4_10_patrones_numero_4.txt')
Matriz_neuronas_5_deformada = np.loadtxt('Ej_4_10_patrones_numero_5.txt')
Matriz_neuronas_6_deformada = np.loadtxt('Ej_4_10_patrones_numero_6.txt')
Matriz_neuronas_7_deformada = np.loadtxt('Ej_4_10_patrones_numero_7.txt')
Matriz_neuronas_8_deformada = np.loadtxt('Ej_4_10_patrones_numero_8.txt')
Matriz_neuronas_9_deformada = np.loadtxt('Ej_4_10_patrones_numero_9.txt')

N = Matriz_neuronas_0_deformada.shape[1]

# Modifico la forma de la matriz para que sea un array de M matrices 30x30
Matriz_neuronas_0_deformada = Matriz_neuronas_0_deformada.reshape(-1, N, N)
Matriz_neuronas_1_deformada = Matriz_neuronas_1_deformada.reshape(-1, N, N)
Matriz_neuronas_2_deformada = Matriz_neuronas_2_deformada.reshape(-1, N, N)
Matriz_neuronas_3_deformada = Matriz_neuronas_3_deformada.reshape(-1, N, N)
Matriz_neuronas_4_deformada = Matriz_neuronas_4_deformada.reshape(-1, N, N)
Matriz_neuronas_5_deformada = Matriz_neuronas_5_deformada.reshape(-1, N, N)
Matriz_neuronas_6_deformada = Matriz_neuronas_6_deformada.reshape(-1, N, N)
Matriz_neuronas_7_deformada = Matriz_neuronas_7_deformada.reshape(-1, N, N)
Matriz_neuronas_8_deformada = Matriz_neuronas_8_deformada.reshape(-1, N, N)
Matriz_neuronas_9_deformada = Matriz_neuronas_9_deformada.reshape(-1, N, N)

M = Matriz_neuronas_0_deformada.shape[0]

# Cargo los patrones originales
patron_original_0 = np.loadtxt('0.txt')
patron_original_1 = np.loadtxt('1.txt')
patron_original_2 = np.loadtxt('2.txt')
patron_original_3 = np.loadtxt('3.txt')
patron_original_4 = np.loadtxt('4.txt')
patron_original_5 = np.loadtxt('5.txt')
patron_original_6 = np.loadtxt('6.txt')
patron_original_7 = np.loadtxt('7.txt')
patron_original_8 = np.loadtxt('8.txt')
patron_original_9 = np.loadtxt('9.txt')

# Calculo el valor de a
def a(patron_original):
    return np.mean(patron_original)

# Defino la función que calcula el solapamiento
def solapamiento(patron_original, s, a, N):

    m = 0.0
    for i in range(N):
        for j in range(N):
                
            m += (patron_original[i, j] - a) * (s[i, j] - a) ;
        
    m = m / ((N * N) * a * (1 - a)) # Pongo el valor absoluto para que no influya que la imagen se ha invertido
    return m

# Calculo el solapamiento para cada configuración de las neuronas
solapamiento_0_deformada = np.zeros(M)
solapamiento_1_deformada = np.zeros(M)
solapamiento_2_deformada = np.zeros(M)
solapamiento_3_deformada = np.zeros(M)
solapamiento_4_deformada = np.zeros(M)
solapamiento_5_deformada = np.zeros(M)
solapamiento_6_deformada = np.zeros(M)
solapamiento_7_deformada = np.zeros(M)
solapamiento_8_deformada = np.zeros(M)
solapamiento_9_deformada = np.zeros(M)

for k in range(M):
    solapamiento_0_deformada[k] = solapamiento(patron_original_0, Matriz_neuronas_0_deformada[k], a(patron_original_0), N)
    solapamiento_1_deformada[k] = solapamiento(patron_original_1, Matriz_neuronas_1_deformada[k], a(patron_original_1), N)
    solapamiento_2_deformada[k] = solapamiento(patron_original_2, Matriz_neuronas_2_deformada[k], a(patron_original_2), N)
    solapamiento_3_deformada[k] = solapamiento(patron_original_3, Matriz_neuronas_3_deformada[k], a(patron_original_3), N)
    solapamiento_4_deformada[k] = solapamiento(patron_original_4, Matriz_neuronas_4_deformada[k], a(patron_original_4), N)
    solapamiento_5_deformada[k] = solapamiento(patron_original_5, Matriz_neuronas_5_deformada[k], a(patron_original_5), N)
    solapamiento_6_deformada[k] = solapamiento(patron_original_6, Matriz_neuronas_6_deformada[k], a(patron_original_6), N)
    solapamiento_7_deformada[k] = solapamiento(patron_original_7, Matriz_neuronas_7_deformada[k], a(patron_original_7), N)
    solapamiento_8_deformada[k] = solapamiento(patron_original_8, Matriz_neuronas_8_deformada[k], a(patron_original_8), N)
    solapamiento_9_deformada[k] = solapamiento(patron_original_9, Matriz_neuronas_9_deformada[k], a(patron_original_9), N)

# Creo un vector que contiene los números del 0 al M-1 para graficar el solapamiento en función del número del paso Monte Carlo
paso_Monte_Carlo = np.arange(M)

# Defino la figura donde voy a graficar
fig, ax = plt.subplots(figsize = (5, 5))

ax.plot(paso_Monte_Carlo, solapamiento_0_deformada, marker = 'o', color = "red", label = "Solapamiento con 0")
ax.plot(paso_Monte_Carlo, solapamiento_1_deformada, marker = 'o', color = "blue", label = "Solapamiento con 1")
ax.plot(paso_Monte_Carlo, solapamiento_2_deformada, marker = 'o', color = "green", label = "Solapamiento con 2")
ax.plot(paso_Monte_Carlo, solapamiento_3_deformada, marker = 'o', color = "orange", label = "Solapamiento con 3")
ax.plot(paso_Monte_Carlo, solapamiento_4_deformada, marker = 'o', color = "purple", label = "Solapamiento con 4")
ax.plot(paso_Monte_Carlo, solapamiento_5_deformada, marker = 'o', color = "brown", label = "Solapamiento con 5")
ax.plot(paso_Monte_Carlo, solapamiento_6_deformada, marker = 'o', color = "gray", label = "Solapamiento con 6")
ax.plot(paso_Monte_Carlo, solapamiento_7_deformada, marker = 'o', color = "pink", label = "Solapamiento con 7")
ax.plot(paso_Monte_Carlo, solapamiento_8_deformada, marker = 'o', color = "yellow", label = "Solapamiento con 8")
ax.plot(paso_Monte_Carlo, solapamiento_9_deformada, marker = 'o', color = "cyan", label = "Solapamiento con 9")

ax.axhline(0.75, linestyle='--', color = "black")

ax.set_title("Solapamiento en función del\n tiempo en pasos Monte Carlo")
ax.set_xlabel("Paso Monte Carlo")
ax.set_ylabel("m")
ax.legend()

fig.savefig("Solapamiento_10_patrones.png", dpi = 300)
plt.show()
```
